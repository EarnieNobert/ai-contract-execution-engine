```
from contract_agents.intake_agent import generate_intake

scope = generate_intake(
    contract_name="Plumbing Chatbot Contract",
    client_name="ABC Plumbing",
    raw_request="Build a customer support chatbot for our plumbing company using our FAQ and service policies.",
    constraints="Use Streamlit\nKeep it simple",
    deadline="2026-05-01",
    notes="Need a strong first draft"
)

scope.model_dump()
```

```
from contract_agents.planner_agent import generate_plan

plan = generate_plan(scope.model_dump())
plan.model_dump()
```

```
from contract_agents.builder_agent import generate_assets

assets = generate_assets(scope.model_dump(), plan.model_dump())
assets
```

```
from contract_agents.qa_agent import run_qa

qa = run_qa(scope.model_dump(), plan.model_dump(), assets)
qa.model_dump()
```
