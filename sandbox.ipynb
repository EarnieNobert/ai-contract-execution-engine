```
from contract_agents.intake_agent import generate_intake

tests = [
    {
        "contract_name": "HVAC Support Chatbot",
        "client_name": "CoolAir HVAC",
        "raw_request": "Build a customer support chatbot for our HVAC company using our FAQ and warranty policies.",
        "constraints": "Use Streamlit\nKeep it lightweight",
        "deadline": "2026-06-01",
        "notes": "Need first draft for demo"
    },
    {
        "contract_name": "Sales Insights Dashboard",
        "client_name": "Northwind Retail",
        "raw_request": "Analyze our monthly sales CSV files and generate business insights and dashboard recommendations.",
        "constraints": "Use pandas\nKeep outputs beginner friendly",
        "deadline": "2026-06-15",
        "notes": "Need a report draft"
    },
    {
        "contract_name": "Python Learning Assistant",
        "client_name": "CodePath Academy",
        "raw_request": "Build an AI tutor that teaches beginner Python concepts and generates quizzes and grading logic.",
        "constraints": "Use Streamlit\nSimple UI",
        "deadline": "2026-07-01",
        "notes": "Need MVP outline"
    }
]

for test in tests:
    result = generate_intake(**test, use_ai=True)
    print(test["contract_name"])
    print("Project type:", result.project_type)
    print("Deliverables:", result.deliverables)
    print("Missing info:", result.missing_info)
    print("-" * 80)
```
